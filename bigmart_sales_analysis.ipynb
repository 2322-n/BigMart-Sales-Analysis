```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("bigmart_excel_cleaned.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df['Item Fat Content'].value_counts()
```

```python
df['Item Fat Content'] = df['Item Fat Content'].replace('low fat', 'Low Fat')
df.head()
```

```python
df['Item Fat Content'].value_counts()
```

```python
df.info()
```

<p style="font-size:20px">1. Sales by Item Type</p>

```python
graph1 = df.groupby('Item Type')['Sales'].sum().sort_values()
# print(graph1)
# plt.bar('Item Type', 'Sales', data=df).graph1
plt.figure(figsize=(30,15))
plt.bar(graph1.index, graph1.values)
plt.xlabel('Total Sales', fontsize=18)
plt.ylabel('Item Type', fontsize=18)
plt.title('Total Sales by Item Type', fontsize=30)
plt.show()
```

```python
graph1 = df.groupby('Item Type')['Sales'].sum().sort_values().reset_index()
# print(graph1)
# plt.bar('Item Type', 'Sales', data=df).graph1
plt.figure(figsize=(30,15))
plt.bar(graph1['Item Type'], graph1['Sales'])
# plt.bar(graph1.index, graph1.values)
plt.xlabel('Total Sales', fontsize=18)
plt.ylabel('Item Type', fontsize=18)
plt.title('Total Sales by Item Type', fontsize=30)
plt.show()
```

<p style="font-size:20px">2. Sales by Outlet Type</p>

```python
graph2 = df.groupby('Outlet Type')['Sales'].sum().sort_values()
# print(graph2)

plt.figure(figsize=(16,7))
plt.bar(graph2.index, graph2.values)
plt.xlabel('Outlet Type', fontsize=16)
plt.ylabel('Sales', fontsize=16)
plt.title('Sales by Outlet Type', fontsize=20)
plt.show()
```

<p style="font-size:20px">3. Rating Distribution</p>

```python
plt.hist(df['Rating'])
plt.title('Rating Distribution', fontsize=20, fontstyle='oblique')
plt.show()
```

<p style="font-size:20px">4. Correlation Heatmap</p>

```python
plt.figure(figsize=(16,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='viridis')
plt.show()
```

<p style="font-size:20px">5. Which Fat Content Category Contributes More to Sales?</p>

```python
graph3 = df.groupby('Item Fat Content')['Sales'].sum()
graph3
```

<p style="font-size:20px">6. Do Older Outlets Perform Better?</p>

```python
graph4 = df.groupby('Outlet Age')['Sales'].sum()
# print(graph4)

plt.figure(figsize=(10,5))
plt.plot(graph4.index, graph4.values, marker='o')
plt.title('Total Sales by Outlet Age')
plt.xlabel('Outlet Age')
plt.ylabel('Sales')
plt.show()
```

<p style="font-size:20px">7. Which Location Type Has the Highest Sales?</p>

```python
graph5 = df.groupby('Outlet Location Type')['Sales'].sum().sort_values(ascending=False)
# print(graph5)

plt.figure(figsize=(8,5))
plt.bar(graph5.index, graph5.values)
plt.title('Total Sales by Outlet Location')
plt.xlabel('Outlet Location')
plt.ylabel('Sales')
plt.show()
```

<p style="font-size:20px">Insights</p>

1. Fruites and Vegetables has the highest sales.
2. Supermarket Type 1 generates maximum revenue.
3. There is no relation of rating with product sales.
4. Outlet age does not have much impact on sales. Outlets aged 8 years perform the best, whereas 15-year-old outlets have the lowest sales.
5. Tier 3 location generates the most sales, whereas tier 1 location generates the least sales.
